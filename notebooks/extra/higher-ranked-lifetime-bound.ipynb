```rust
trait Handler {
    fn handle(&self, input: &str);
}

impl<F> Handler for F
where
    F: for<'a> Fn(&'a str) + Send + Sync + 'static,
{
    fn handle(&self, input: &str) {
        self(input);
    }
}
```

```rust
struct Registry {
    handlers: std::collections::HashMap<String, Box<dyn Handler>>,
}

impl Registry {
    fn new() -> Self {
        Registry {
            handlers: std::collections::HashMap::new(),
        }
    }   

    fn register<F>(&mut self, name: &str, handler: F)
    where F: for<'a> Fn(&'a str) + Send + Sync + 'static, {
        self.handlers.insert(name.to_string(), Box::new(handler));
    }

    fn call(&self, name: &str, input: &str) {
        if let Some(handler) = self.handlers.get(name) {
            handler.handle(input);
        } else {
            println!("No handler registered for {}", name);
        }
    }
}
```

```rust
fn main() {
    let mut registry = Registry::new();

    registry.register("print", |input: &str| {
        println!("Handler called with input: {}", input);
    });

    registry.call("print", "Hello, world!");
    registry.call("unknown", "This will not be handled.");
}

main();
```
